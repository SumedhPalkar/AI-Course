#### 8.	What is correlation?

#### 9.	What is multicollinearity?

- Multicollinearity is a condition in which two or more independent variables in dataset are highly correlated with each other. 
- Multicollinearity = High correlation between independent variables

#### 10.	How to detect multicollinearity?

1. Correlation Matrix
2. Variance Inflation Factor (VIF)
3. Torelance 
4. Condition Index (CI)
5. Regression Symptoms 

Among these, VIF is the most widely used measure. A VIF value greater than 10 or a tolerance value below 0.1 indicates serious multicollinearity among the independent variables.

#### 11.	What are the remedies for multicollinearity?

* **Do Nothing:** If the Correlation is not that extreme, we can ignore it. If the correlated variables are not used in solving our business question, they can be ignored.
* **Remove One Variable**: Like in dummy variable trap
* **Combine the correlated variables:** Like creating a seniority score based on Age and Years of experience
* Principal Component Analysis

#### 12.	What is the R-Squared Statistics?

- R-squared is a statistical measure that tells us how much variation in the dependent variable is explained by the independent variables in a regression model.
- it represents how much of our data is being explained by our model. 



- Where, RSS: Residual Sum of squares  & TSS: Total sum of squares

#### 13.	What is an adjusted R-Squared Statistics?

- It is a modified version of R-squared that adjust the numbers of independent variables in a regression model. 
- It increases only when adding a variable improves the model sufficiently, otherwise it decreases. 

#### 14.	Why do we use adj R-squared?

- We use Adjusted R-squared to evaluate how well a regression model explains the variation in the dependent variable while adjusting for the number of independent variables, helping prevent unnecessary variables from making the model appear better.

#### 15.	Why adj R-squared decreases when we use incompetent variables?

- It penalizes unnecessary predictors that do not sufficiently improves the explanatory power of the regression model.

#### 19.	What are Bias and Variance? What is Bias Variance Trade-off?

**Bias**: 
- Bias is a error caused by a model making overly simple assumptions about the data.
- It is a measure of polarity of the results.
- High bias means model is too simple.

**Variance**: 
- Variance measures how much a model's prediction changes when trained on different dataset.
- It is a measure of degree of spread of predicted results.
- High variance means the model is too sensitive to training data.

**Bias Variance Trade-off**:
- It is the balance between bias and variance to buils a model that performs well on both training data and unseen data.

#### 20.	What is the null and alternate hypothesis?

**Null Hypothesis**: Its a default assumption that there is no effect, no difference, or no relationship between variables.

**Alternate Hypothesis**: Its a statement that there is an effect, difference and relationship between variables.

#### 24.	How to handle categorical values in the data?

- Categorical variables cannot directly used by most machines learning algorithms. 
- They are converted into numerical values using techniques such as label encoding, one-hot encoding, dummy variable encoding, ordinal encoding and target encoding. 
- The choice depends on whether the categories are ordinal or nominal and the type of model being used. 

#### 25.	What is regularization, and why do we need it?

- Its a technique used to prevent overfitting by adding a penalty term to the loss function, which discourages large model coefficients. 
- overfitting occurs when model learn not only the real patterns in the training data but also the noice and random fluctuations.

#### 26.	Explain Ridge Regression.

- Its a modified version of standard linear regression that uses L2 regularization to prevent overfitting and handle multicollinearity.
- It helps prevent overfitting by adding a penalty to large coefficient values. 

#### 27.	Explain Lasso Regression.

- Lasso regression is a regularized linear regression technique that uses L1 regularization to reduce overfitting and automatically select important features by shrinking some coefficient exactly to zero.

#### 28.	Explain Elastic Net.

- Elastic Net is a middle ground between Ridge Regression and Lasso Regression.
- It combines both L1 (Lasso) and L2 (Ridge) regularization to reduce overfitting, handle multicollinearity and perform features selection. 

#### 29.	Why do we do a train test split?

- We perform a train split to evaluate a model's ability to generalize to unseen data, detect overfitting, and obtain an unbiased estimate of its performance. 


#### 30.	What is polynomial regression? When to use it?

- Polynomial Regression can be defined as a mechanism to predict a _dependent variable_ based on the polynomial relationship with the _independent variable_.
- It is used when the relationship between the independent and dependent variables is curved rather than linear.
